## Series in Pandas

Series in pandas may seem like an array style stuff but it isnt, its more similar as a weird ordered dictionary

```python
g7_arr = np.array([35.467, 63.951, 80.49, 127.061, 64.511, 318.523, 345.23])
g7_pop = pd.Series(g7_arr)
```

```python
g7_pop
```

```python
g7_pop.name = "G7 population in millions"
```

```python
g7_pop.values
```

```python
type(g7_pop.values)
```

```python
g7_pop.index
```

```python
g7_pop[0]
```

### Indexing

```python
g7_pop.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States'
]
```

```python
g7_pop
```

Now I dont have to remember what numerical index belongs to what country

```python
g7_pop['Canada']
```

Indexes location can be used also with iloc attribute

```python
g7_pop.iloc[0]
```

```python
g7_pop.iloc[[0, 1, 4]]
```

*** You can also use slicing numpy-like w pandas (Last element is INCLUSIVE)***

```python
g7_pop['Canada':'Italy']
```

### Conditional Selection (Just like with numpy)

```python
g7_pop
```

```python
g7_pop > 70
```

```python
g7_pop[g7_pop > 80]
```

```python
g7_pop.mean()
```

```python
g7_pop[g7_pop > g7_pop.mean()]
```

```python
g7_pop.std()
```

```python
g7_pop * 1_000_000
```

```python
np.log(g7_pop)
```

```python
g7_pop[g7_pop > 70] = 90
```

```python
g7_pop
```
