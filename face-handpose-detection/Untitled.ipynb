### Install and import Dependencies

```python
# Importaciones
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
import cv2
import pygame
```

```python
pygame.mixer.init()

# Lista de sonidos
sounds = [
    pygame.mixer.Sound("./sounds/e.mp3"), # Left Index
    pygame.mixer.Sound("./sounds/d#d.mp3"), # Middle Left
    pygame.mixer.Sound("./sounds/d#.mp3"), # Anular Left
    pygame.mixer.Sound("./sounds/c#d.mp3"), #Indice Right
    pygame.mixer.Sound("./sounds/c#.mp3"), # Middle Right
    pygame.mixer.Sound("./sounds/e.mp3") # Anular Right
]
```

```python
# Function to detect finger down
def is_finger_down(landmarks, finger_tip, finger_mcp):
    return landmarks[finger_tip].y > landmarks[finger_mcp].y
```

```python
# Model Setup
MODEL_PATH = "hand_landmarker.task"

base_options = python.BaseOptions(model_asset_path=MODEL_PATH)
options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=2
)

detector = vision.HandLandmarker.create_from_options(options)

# Camera Live Recording
cap = cv2.VideoCapture(0)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Para que no se vea volteado la foto de la camara
    frame = cv2.flip(frame, 1)

    # Conversion a rgb
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = vision.MPImage(image_format=vision.ImageFormat.SRGB, data=rgb_frame)

    result = detector.detect(mp_image)

    if result.hand_landmarks:
        print("Hands detected:", len(result.hand_landmarks))

    cv2.imshow("MediaPipe Tasks", frame)
    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
