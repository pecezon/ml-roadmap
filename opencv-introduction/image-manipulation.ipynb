```python
import pandas as pd
import numpy as np

from glob import glob

import cv2
import matplotlib.pylab as plt
```

```python
# Read images

dog_files = glob('./test_set/dogs/*.jpg')
cat_files = glob('./test_set/cats/*.jpg')
```

```python
# Reading images as 3 dimension numpy arrays
img_mpl = plt.imread(cat_files[20])
img_cv2 = cv2.imread(cat_files[20])

# Height, Width, Channels array
```

```python
# Display images
fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(img_mpl)
plt.show()
```

```python
# Image channels display rgb channels cv3 reads in channels BGR
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(img_mpl[:,:,0], cmap='Reds')
axs[1].imshow(img_mpl[:,:,1], cmap='Greens')
axs[2].imshow(img_mpl[:,:,2], cmap='Blues')
```

```python
# Image channels display rgb channels cv3 reads in channels BGR
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].imshow(img_cv2)
axs[1].imshow(img_mpl)
axs[0].axis("off")
axs[1].axis("off")
```

```python
# Convert from BGR to RGB for CV2
img_cv2_rgb = cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB)
fig, ax = plt.subplots()
ax.imshow(img_cv2_rgb)
ax.axis('off')
plt.show()
```

### Image manipulation

```python
img = plt.imread(dog_files[5])
fig, ax = plt.subplots()
ax.imshow(img)
ax.axis('off')
plt.show()
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
img_gray.shape
fig, axs = plt.subplots(figsize=(10, 5))
axs.imshow(img_gray, cmap="Greys")
plt.show()
```

```python
# Resizing and Scaling
img_resized = cv2.resize(img, None, fx=0.25, fy=0.25)
fig, axs = plt.subplots(figsize=(10, 5))
axs.imshow(img_resized)
plt.show()
```

```python
img_resized = cv2.resize(img, (5000, 5000), interpolation=cv2.INTER_CUBIC)
fig, axs = plt.subplots(figsize=(10, 5))
axs.imshow(img_resized)
plt.show()
img_resized.shape
```

```python
# Kernels

#Sharpen Image
kernel_sharpening = np.array([[-1,-1, -1],
                             [-1,9, -1],
                             [-1,-1, -1]])
sharpened = cv2.filter2D(img, -1, kernel_sharpening)

fig, axs = plt.subplots(figsize=(10, 5))
axs.imshow(sharpened)
plt.show()
```

```python
# Bluring
kernel_3x3 = np.ones((3, 3), np.float32)/9
blured = cv2.filter2D(img, -1, kernel_3x3)

fig, axs = plt.subplots(figsize=(10, 5))
axs.imshow(blured)
plt.show()
```
