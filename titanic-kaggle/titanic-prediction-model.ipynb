### Training Table

```python
train = pd.read_csv('train.csv')
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier(n_estimators=200,
    random_state=42 
                            )
# Take sex in account
train["Sex"] = train["Sex"].map({"male": 0, "female": 1})
test["Sex"] = test["Sex"].map({"male": 0, "female": 1})

features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare"]

X = train[features].fillna(train[features].median())
Y = train["Survived"]

clf.fit(X, Y)

testX = test[features].fillna(test[features].median())

predict = clf.predict(testX)
```

```python
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": predict
})

submission.to_csv("randforest_submission.csv", index=False)
```

### Gradient Boost

```python
from sklearn.ensemble import GradientBoostingClassifier
gb = GradientBoostingClassifier(n_estimators=200, random_state=1, learning_rate=0.2)
gb.fit(X, Y)
gb_predict = gb.predict(testX)
```

```python
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": gb_predict
})

submission.to_csv("gb_submission.csv", index=False)
```

### XGBoost

```python
from xgboost import XGBClassifier

bst = XGBClassifier(n_estimators=2, max_depth=2, learning_rate=1, objective='binary:logistic')
bst.fit(X, Y)
bst_predict = bst.predict(testX)
```

```python
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": bst_predict
})

submission.to_csv("bst_submission.csv", index=False)
```
